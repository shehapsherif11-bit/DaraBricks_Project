# The Transformation Logic

```python
query = """
SELECT
    sd.order_number,
    pr.product_key,     -- الكي المتشفر النظيف اللي هيروح لـ Power BI
    cu.customer_key,    -- كي العميل المتشفر
    sd.order_date,
    sd.ship_date,
    sd.due_date,
    sd.sales_amount,
    sd.sales_quantity,
    sd.sales_price
FROM workspace.silver.crm_sales sd
LEFT JOIN workspace.gold.dim_products pr
    -- نربط الكود القديم في المبيعات بالكود القديم في المنتجات
    ON cast(sd.product_key as string) = cast(pr.product_number as string)
LEFT JOIN workspace.gold.dim_customers cu
    ON sd.customer_id = cu.customer_id;
"""

df = spark.sql(query)
df.limit(10).display()
```

# Writing Gold Table

```python
df.write.mode("overwrite").format("delta").saveAsTable("workspace.gold.fact_sales")
```
